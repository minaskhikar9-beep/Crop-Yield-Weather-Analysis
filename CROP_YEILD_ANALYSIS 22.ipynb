## 7. Data Cleaning

Data cleaning is an important step in data analysis. It helps improve data quality by identifying missing values, duplicate records, and incorrect data types.

In this project, data cleaning is performed to make the agricultural dataset suitable for analysis and visualization.

```python
# Check missing values
print("Missing Values:")
display(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Display dataset shape after cleaning
print("\nDataset Shape After Removing Duplicates:")
print(df.shape)
```

## Identify Important Columns (Code)

```python
# Automatically detect common column names

def find_column(possible_names):
    for col in df.columns:
        for name in possible_names:
            if name.lower() in col.lower():
                return col
    return None

year_col = find_column(["year"])
state_col = find_column(["state"])
district_col = find_column(["district"])
crop_col = find_column(["crop"])
yield_col = find_column(["yield", "productivity"])
production_col = find_column(["production"])
rainfall_col = find_column(["rainfall", "precipitation"])
temperature_col = find_column(["temperature", "temp"])

print("Year Column:", year_col)
print("State Column:", state_col)
print("District Column:", district_col)
print("Crop Column:", crop_col)
print("Yield Column:", yield_col)
print("Production Column:", production_col)
print("Rainfall Column:", rainfall_col)
print("Temperature Column:", temperature_col)
```

## Convert Numeric Columns (Code)

```python
# Convert detected numeric columns to numeric format

numeric_columns = [
    year_col,
    yield_col,
    production_col,
    rainfall_col,
    temperature_col
]

for col in numeric_columns:
    if col is not None:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("Numeric columns converted successfully.")

display(df.head())
```

## 9. Year-wise Crop Yield Analysis

Year-wise analysis helps us understand how crop yield or production changes over time.

A line chart is used to visualize yearly agricultural trends. This analysis helps identify increases, decreases, and fluctuations in the recorded crop outcomes.

```python
# Select yield if available; otherwise use production
outcome_col = yield_col if yield_col is not None else production_col

if year_col is not None and outcome_col is not None:

    yearly_data = df.dropna(
        subset=[year_col, outcome_col]
    )

    yearly_summary = yearly_data.groupby(
        year_col
    )[outcome_col].mean().sort_index()

    display(yearly_summary.to_frame("Average Crop Outcome"))

    plt.figure(figsize=(12, 5))

    plt.plot(
        yearly_summary.index,
        yearly_summary.values,
        marker="o"
    )

    plt.title("Year-wise Crop Yield / Production Analysis")
    plt.xlabel("Year")
    plt.ylabel(f"Average {outcome_col}")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

else:
    print("Year or crop yield/production column not found.")
```

## 10. Crop-wise Analysis

Crop-wise analysis compares the average yield or production of different crops.

This analysis helps identify which crops have higher average recorded outcomes in the dataset. The results depend on the measurement used and should not be compared directly when crops use different units or scales.

## Crop-wise Visualization (Code)

```python
if crop_col is not None and outcome_col is not None:

    crop_summary = df.groupby(
        crop_col
    )[outcome_col].mean().dropna().sort_values(
        ascending=False
    )

    print("Top 10 Crops by Average Outcome:")
    display(crop_summary.head(10).to_frame("Average Outcome"))

    crop_summary.head(10).sort_values().plot(
        kind="barh",
        figsize=(10, 6)
    )

    plt.title("Top 10 Crops by Average Yield / Production")
    plt.xlabel(f"Average {outcome_col}")
    plt.ylabel("Crop")
    plt.tight_layout()
    plt.show()

else:
    print("Crop or outcome column not found.")
```

##  13. Rainfall Analysis

Rainfall is an important weather factor affecting agricultural production. Insufficient rainfall may create water shortages, while excessive rainfall can cause flooding and crop damage.

This analysis examines the relationship between recorded rainfall and crop yield or production using a scatter plot and correlation coefficient.

## Rainfall vs Crop Yield (Code)

```python
rainfall_col = "Rain_Ideal (mm/day)_Ideal"
outcome_col = "PRODUCTION (1000 tons)"

if rainfall_col is not None and outcome_col is not None:

    rainfall_data = df[
        [rainfall_col, outcome_col]
    ].dropna()

    if len(rainfall_data) > 1:

        plt.figure(figsize=(8, 5))

        plt.scatter(
            rainfall_data[rainfall_col],
            rainfall_data[outcome_col],
            alpha=0.6
        )

        plt.title("Rainfall vs Crop Yield / Production")
        plt.xlabel("Rainfall")
        plt.ylabel(f"{outcome_col}")
        plt.tight_layout()
        plt.show()

        correlation = rainfall_data.corr().iloc[0, 1]

        print(
            "Rainfall and Crop Outcome Correlation:",
            round(correlation, 3)
        )

    else:
        print("Not enough complete rainfall records.")

else:
    print("Rainfall or crop outcome column not found.")
```

##  14. Temperature Analysis

Temperature influences crop growth, flowering, and maturity. Very high or low temperatures can affect agricultural productivity depending on the crop and its growth stage.

This analysis explores the association between recorded temperature and crop yield or production.

## Temperature vs Crop Yield (Code)

```python
temperature_col = "T2M_MAX"
outcome_col = "PRODUCTION (1000 tons)"

if temperature_col is not None and outcome_col is not None:

    temperature_data = df[
        [temperature_col, outcome_col]
    ].dropna()

    if len(temperature_data) > 1:

        plt.figure(figsize=(8, 5))

        plt.scatter(
            temperature_data[temperature_col],
            temperature_data[outcome_col],
            alpha=0.6
        )

        plt.title("Temperature vs Crop Yield / Production")
        plt.xlabel("Temperature")
        plt.ylabel(f"{outcome_col}")
        plt.tight_layout()
        plt.show()

        correlation = temperature_data.corr().iloc[0, 1]

        print(
            "Temperature and Crop Outcome Correlation:",
            round(correlation, 3)
        )

    else:
        print("Not enough complete temperature records.")

else:
    print("Temperature or crop outcome column not found.")
```

## 15. Correlation and Covariance Analysis

Correlation measures the strength and direction of the relationship between two numerical variables. Its value ranges from -1 to +1.

Covariance indicates whether two variables tend to increase or decrease together.

These techniques help examine relationships between rainfall, temperature, and crop outcomes. However, correlation does not prove that one variable causes changes in another.

```python
numeric_data = df.select_dtypes(include="number")

if numeric_data.shape[1] >= 2:

    correlation_matrix = numeric_data.corr()

    print("Correlation Matrix:")
    display(correlation_matrix)

    print("Covariance Matrix:")
    display(numeric_data.cov())

    plt.figure(figsize=(10, 7))

    plt.imshow(
        correlation_matrix,
        aspect="auto",
        vmin=-1,
        vmax=1
    )

    plt.colorbar(label="Correlation")

    plt.xticks(
        range(len(correlation_matrix.columns)),
        correlation_matrix.columns,
        rotation=90
    )

    plt.yticks(
        range(len(correlation_matrix.columns)),
        correlation_matrix.columns
    )

    plt.title("Correlation Matrix of Crop and Weather Variables")
    plt.tight_layout()
    plt.show()

else:
    print("At least two numeric columns are required.")
```

## 16. Outlier Detection

Outliers are observations that differ significantly from the general pattern of a dataset. In agricultural data, outliers may result from extreme weather conditions, unusual production values, measurement errors, or genuine regional differences.

The Interquartile Range (IQR) method is used to identify unusually low or high values. Detected outliers should be investigated rather than automatically removed.

```python
if outcome_col is not None:

    data = df[outcome_col].dropna()

    if not data.empty:

        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)

        IQR = Q3 - Q1

        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR

        outliers = df[
            (df[outcome_col] < lower_limit) |
            (df[outcome_col] > upper_limit)
        ]

        print("Lower Limit:", lower_limit)
        print("Upper Limit:", upper_limit)
        print("Number of Outliers:", len(outliers))

        display(outliers.head(20))

        plt.figure(figsize=(8, 4))
        plt.boxplot(data)
        plt.title("Box Plot for Crop Yield / Production")
        plt.ylabel(outcome_col)
        plt.tight_layout()
        plt.show()

    else:
        print("No valid outcome values found.")

else:
    print("Crop yield or production column not found.")
```
